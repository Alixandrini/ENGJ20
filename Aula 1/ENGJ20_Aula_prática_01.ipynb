## 📐 Fundamentação Teórica

Na prática de campo (topografia e geodésia), é muito comum encontrarmos coordenadas no formato sexagesimal (**Graus, Minutos e Segundos**):

$$\text{Graus Decimais (DD)} = \text{Graus} + \frac{\text{Minutos}}{60} + \frac{\text{Segundos}}{3600}$$

> **Atenção aos Sinais:**  
> Para latitudes no **Hemisfério Sul (S)** ou longitudes a **Oeste de Greenwich (W)**, o valor final em Graus Decimais deve ser **negativo**.



```
# ==============================================================================
# BLOCO 1: Tipos de Dados Primitivos e Variáveis
# Execute esta célula clicando no botão 'Play' ou pressionando Shift + Enter
# ==============================================================================

# Declaração de variáveis para a Estação Geodésica 'VT01'
nome_estacao = "VT01"          # Tipo String (texto)
graus = 12                      # Tipo Int (inteiro)
minutos = 58                    # Tipo Int (inteiro)
segundos = 22.8                 # Tipo Float (número decimal)
hemisferio = "S"               # Tipo String (texto)

# Exibindo os valores e os tipos de dados
print(f"Estação: {nome_estacao} (Tipo: {type(nome_estacao)})")
print(f"Graus: {graus} (Tipo: {type(graus)})")
print(f"Minutos: {minutos} (Tipo: {type(minutos)})")
print(f"Segundos: {segundos} (Tipo: {type(segundos)})")
```
