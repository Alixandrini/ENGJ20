## 📋 BLOCO 1 — Listas
### ⏱️ ~20 minutos

Uma **lista** é uma sequência **ordenada** e **mutável** de elementos.

```python
lista = [elemento0, elemento1, elemento2, ...]
```

- Aceita elementos de **tipos diferentes**
- **Ordenada**: a posição importa
- **Mutável**: podemos alterar, adicionar ou remover elementos
- Indexação começa em **0** (ou -1 para o último)

```
 Índice positivo:   0      1      2      3      4
                 ┌──────┬──────┬──────┬──────┬──────┐
  lista        = │PT-01 │PT-02 │PT-03 │PT-04 │PT-05 │
                 └──────┴──────┴──────┴──────┴──────┘
 Índice negativo:  -5     -4     -3     -2     -1
```


````python
# ── 1.1  Criação de listas e indexação ───────────────────────────────────────
# Pontos de uma poligonal levantada em campo

nomes_pts = ["PT-01", "PT-02", "PT-03", "PT-04", "PT-05"]
este_utm  = [683450.0, 683920.5, 684100.3, 683750.8, 683300.2]   # metros
norte_utm = [7456200.0, 7456650.3, 7456300.7, 7455900.1, 7455700.5]
altitude  = [842.3, 851.1, 847.6, 839.2, 835.8]                   # metros

print("=== Lista de pontos levantados ===")
print(f"Total de pontos     : {len(nomes_pts)}")
print(f"Primeiro ponto      : {nomes_pts[0]}")
print(f"Ultimo ponto        : {nomes_pts[-1]}")
print(f"Terceiro ponto      : {nomes_pts[2]}")
print(f"Ante-penultimo      : {nomes_pts[-3]}")
print()
print(f"Este do PT-01       : {este_utm[0]} m")
print(f"Norte do ultimo pt  : {norte_utm[-1]} m")
````

````python
# ── 1.2  Fatiamento (Slicing) ────────────────────────────────────────────────
# sintaxe: lista[inicio : fim : passo]
#   - inicio  : índice onde começa (inclusivo)
#   - fim     : índice onde termina (exclusivo)
#   - passo   : de quantos em quantos

print("Fatiamento de listas (slicing):")
print()
print("nomes_pts[1:4]   =", nomes_pts[1:4])      # índices 1, 2, 3
print("nomes_pts[:3]    =", nomes_pts[:3])        # do inicio ate 2
print("nomes_pts[2:]    =", nomes_pts[2:])        # do índice 2 ao fim
print("nomes_pts[::2]   =", nomes_pts[::2])       # de 2 em 2
print("nomes_pts[::-1]  =", nomes_pts[::-1])      # invertida
print()
# Útil: selecionar apenas os pontos do meio (excluindo início e fim)
pts_intermediarios = nomes_pts[1:-1]
print(f"Pontos intermediarios: {pts_intermediarios}")
````

````python
# ── 1.3  Métodos essenciais de lista ─────────────────────────────────────────
import copy

pontos = copy.copy(nomes_pts)   # cópia independente
print("Lista original:", pontos)
print()

# append — adiciona ao final
pontos.append("PT-06")
print("Depois de append('PT-06'):", pontos)

# insert — adiciona em posição específica
pontos.insert(2, "PT-02B")
print("Depois de insert(2,'PT-02B'):", pontos)

# remove — remove pelo valor (primeira ocorrência)
pontos.remove("PT-02B")
print("Depois de remove('PT-02B'):", pontos)

# pop — remove e retorna pelo índice (padrão: último)
removido = pontos.pop()
print(f"pop() removeu: {removido}  →  lista: {pontos}")

# sort — ordena no lugar
desordenado = ["PT-03", "PT-01", "PT-05", "PT-02", "PT-04"]
desordenado.sort()
print("Ordenada:", desordenado)

# reverse — inverte no lugar
desordenado.reverse()
print("Invertida:", desordenado)
````

````python
# ── 1.4  Operações úteis com listas numéricas ────────────────────────────────
print("=== Estatísticas das altitudes ===")
print(f"Valores      : {altitude}")
print(f"Mínima       : {min(altitude):.1f} m  ->  {nomes_pts[altitude.index(min(altitude))]}")
print(f"Máxima       : {max(altitude):.1f} m  ->  {nomes_pts[altitude.index(max(altitude))]}")
print(f"Soma         : {sum(altitude):.1f} m")
print(f"Media        : {sum(altitude)/len(altitude):.2f} m")
print()

# sorted() — retorna uma NOVA lista ordenada (não altera a original)
alt_ordenada = sorted(altitude)
print(f"Altitudes ordenadas (crescente): {alt_ordenada}")
print(f"Original inalterada            : {altitude}")
````

````python
# ── 1.5  List Comprehension — forma compacta e elegante ──────────────────────
# Sintaxe: [expressão for item in iterável if condição]

# Converter altitudes de metros para pés (1 m = 3.28084 ft)
altitudes_pes = [round(h * 3.28084, 2) for h in altitude]
print("Altitudes em pes:", altitudes_pes)

# Filtrar apenas pontos com altitude acima de 840 m
pts_altos = [nomes_pts[i] for i in range(len(altitude)) if altitude[i] > 840]
print("Pontos acima de 840 m:", pts_altos)

# Criar identificadores numerados automaticamente
ids = [f"ID{i+1:03d}" for i in range(len(nomes_pts))]
print("IDs gerados:", ids)
````

````python
# ── 1.6  zip() — percorrer múltiplas listas ao mesmo tempo ───────────────────
print(f"{'Ponto':<8} {'Este (m)':<14} {'Norte (m)':<14} {'Alt (m)':<10}")
print("-" * 50)
for nome, e, n, h in zip(nomes_pts, este_utm, norte_utm, altitude):
    print(f"{nome:<8} {e:<14.2f} {n:<14.2f} {h:<10.1f}")
````

## 📌 BLOCO 2 — Tuplas
### ⏱️ ~20 minutos

Uma **tupla** é uma sequência **ordenada** e **imutável** de elementos.

```python
tupla = (elemento0, elemento1, elemento2)
```

### Diferença fundamental entre Tupla e Lista

| Característica | Lista `[]` | Tupla `()` |
|----------------|-----------|-----------|
| Ordenada | ✅ | ✅ |
| Mutável | ✅ | ❌ |
| Velocidade | Menor | Maior |
| Uso típico | Dados que mudam | Dados fixos / constantes |

> **Regra de ouro na Cartografia:** use tupla para uma **coordenada** (lat, lon não se separam)
> e lista para um **conjunto de pontos** (pode crescer ou mudar).


````python
# ── 2.1  Criação e acesso ─────────────────────────────────────────────────────

# Coordenada de um ponto: (Este, Norte, Altitude)  — dados que não mudam
coord_marco_zero = (683450.0, 7456200.0, 842.3)

print("Coordenada do Marco Zero (E, N, h):")
print(f"  Este     : {coord_marco_zero[0]:.2f} m")
print(f"  Norte    : {coord_marco_zero[1]:.2f} m")
print(f"  Altitude : {coord_marco_zero[2]:.1f} m")
print(f"  Tipo     : {type(coord_marco_zero)}")
print()

# Parâmetros do elipsoide GRS80 (usados no SIRGAS2000) — constantes físicas
grs80 = (6_378_137.0, 298.257222101)   # (semieixo maior a, achatamento inverso 1/f)
print(f"Elipsoide GRS80: a = {grs80[0]:.3f} m  |  1/f = {grs80[1]}")
````

````python
# ── 2.2  Imutabilidade — tentativa de modificar gera erro ────────────────────

try:
    coord_marco_zero[0] = 999.0   # tenta mudar o Este
except TypeError as e:
    print(f"ERRO esperado: {e}")
    print("Tuplas sao imutaveis — dados protegidos de alteracao acidental!")
````

````python
# ── 2.3  Desempacotamento (unpacking) ────────────────────────────────────────
# Extrair cada componente em variáveis separadas — muito usado em Cartografia!

este, norte, alt = coord_marco_zero
print(f"Este     = {este:.2f} m")
print(f"Norte    = {norte:.2f} m")
print(f"Altitude = {alt:.1f} m")
print()

# Desempacotamento com * (captura o "resto")
primeiro, *meio, ultimo = (10.0, 20.5, 30.1, 40.7, 50.2)
print(f"Primeiro: {primeiro}  |  Meio: {meio}  |  Ultimo: {ultimo}")
````

````python
# ── 2.4  Lista de tuplas — estrutura clássica para pontos geoespaciais ────────

# Cada ponto é uma tupla (nome, Este, Norte, Altitude)
poligonal = [
    ("PT-01", 683450.0, 7456200.0, 842.3),
    ("PT-02", 683920.5, 7456650.3, 851.1),
    ("PT-03", 684100.3, 7456300.7, 847.6),
    ("PT-04", 683750.8, 7455900.1, 839.2),
    ("PT-05", 683300.2, 7455700.5, 835.8),
]

print("Poligonal levantada:")
print(f"{'#':<4} {'Ponto':<8} {'E (m)':<14} {'N (m)':<14} {'h (m)':<8}")
print("-" * 52)
for i, (nome, e, n, h) in enumerate(poligonal, 1):
    print(f"{i:<4} {nome:<8} {e:<14.2f} {n:<14.2f} {h:<8.1f}")
````

````python
# ── 2.5  Ordenar lista de tuplas por um critério ──────────────────────────────
import math

# Ordenar pontos por altitude (do mais alto para o mais baixo)
por_altitude = sorted(poligonal, key=lambda pt: pt[3], reverse=True)

print("Pontos ordenados por altitude (decrescente):")
for nome, e, n, h in por_altitude:
    print(f"  {nome}  h = {h:.1f} m")

print()
# Calcular distância ao centroide da poligonal
c_e = sum(p[1] for p in poligonal) / len(poligonal)
c_n = sum(p[2] for p in poligonal) / len(poligonal)
print(f"Centroide da poligonal: E={c_e:.2f}  N={c_n:.2f}")

print()
print("Distancias ao centroide:")
for nome, e, n, _ in poligonal:
    d = math.sqrt((e - c_e)**2 + (n - c_n)**2)
    print(f"  {nome}: {d:.2f} m")
````

````python
# ── 2.6  Tupla como retorno de funcao ─────────────────────────────────────────

def utm_para_geografica(este, norte, fuso=23, hemisferio="S"):
    '''Conversao simplificada UTM -> Geografica (aproximacao didatica).
    Para uso real utilize a biblioteca pyproj.
    '''
    # Meridiano Central do fuso
    mc = (fuso - 1) * 6 - 177
    # Aproximacao linear simples para fins didaticos
    lon_aprox = mc + (este - 500_000) / (111_320 * math.cos(math.radians(-15)))
    lat_aprox = (norte - 10_000_000) / 110_574 if hemisferio == "S" else norte / 110_574
    return round(lat_aprox, 6), round(lon_aprox, 6)   # retorna TUPLA


lat_aprox, lon_aprox = utm_para_geografica(683450.0, 7456200.0)
print(f"Coordenada geografica aproximada: {lat_aprox}°  {lon_aprox}°")
print(f"Tipo do retorno: {type(utm_para_geografica(0, 0))}")
````

## 🔵 BLOCO 3 — Sets (Conjuntos)
### ⏱️ ~20 minutos

Um **set** é uma coleção **não ordenada**, **sem elementos duplicados** e **mutável**.

```python
conjunto = {elemento1, elemento2, elemento3}
```

### Características

| Característica | Set `{}` |
|----------------|---------|
| Ordenado | ❌ (sem índice) |
| Duplicatas | ❌ (automaticamente removidas) |
| Mutável | ✅ |
| Busca | ✅ Muito rápida |

### Operações de Conjuntos (álgebra de conjuntos)

```
A | B   →  União            (tudo de A e de B)
A & B   →  Interseção       (apenas o que está em ambos)
A - B   →  Diferença        (o que está em A mas não em B)
A ^ B   →  Diferença simétrica  (o que está em um, mas não nos dois)
```

> **Aplicação Cartográfica:** verificar quais datums, fusos ou municípios aparecem
> em um conjunto de dados — sem contar duplicatas.


````python
# ── 3.1  Criação de sets ──────────────────────────────────────────────────────

# Datums utilizados em diferentes levantamentos de um projeto
datums_lev1 = {"SAD69", "WGS84", "SIRGAS2000", "SAD69"}     # SAD69 duplicado
datums_lev2 = {"WGS84", "SIRGAS2000", "Corrego Alegre"}

print("Levantamento 1 (raw) : SAD69, WGS84, SIRGAS2000, SAD69")
print("Set criado           :", datums_lev1)    # duplicata removida!
print("Tipo                 :", type(datums_lev1))
print()
print("Levantamento 2       :", datums_lev2)
````

````python
# ── 3.2  Operacoes de conjuntos ───────────────────────────────────────────────
A = datums_lev1
B = datums_lev2

print("A  =", A)
print("B  =", B)
print()
print("A | B  (uniao)              :", A | B)
print("A & B  (intersecao)         :", A & B)
print("A - B  (so em A)            :", A - B)
print("B - A  (so em B)            :", B - A)
print("A ^ B  (dif. simetrica)     :", A ^ B)
print()
print("'WGS84' esta em A?          :", "WGS84" in A)
print("'Corrego Alegre' esta em A? :", "Corrego Alegre" in A)
````

````python
# ── 3.3  Caso real: municipios cobertos por diferentes cartas topograficas ────

# Municípios contidos em cada carta 1:50.000
carta_SF23YA = {"Campinas", "Indaiatuba", "Salto", "Itu", "Valinhos"}
carta_SF23YB = {"Salto", "Itu", "Sorocaba", "Porto Feliz", "Laranjal"}
carta_SF23YC = {"Itu", "Sorocaba", "Tatui", "Boituva", "Porto Feliz"}

todos_municipios    = carta_SF23YA | carta_SF23YB | carta_SF23YC
municipios_comuns   = carta_SF23YA & carta_SF23YB & carta_SF23YC
apenas_na_A         = carta_SF23YA - carta_SF23YB - carta_SF23YC

print(f"Total de municipios cobertos       : {len(todos_municipios)}")
print(f"Municipios cobertos                : {sorted(todos_municipios)}")
print()
print(f"Municipios em todas as 3 cartas    : {municipios_comuns}")
print(f"Municipios exclusivos da carta A   : {apenas_na_A}")
print()
# Verificar se um municipio especifico e coberto
municipio_busca = "Valinhos"
print(f"'{municipio_busca}' e coberto por alguma carta? {municipio_busca in todos_municipios}")
````

````python
# ── 3.4  Sets para remover duplicatas de uma lista ────────────────────────────
# Em campo, o operador pode ter registrado o mesmo ponto duas vezes

pontos_campo = [
    "PT-01", "PT-02", "PT-03", "PT-01",   # PT-01 duplicado!
    "PT-04", "PT-02", "PT-05", "PT-06",   # PT-02 duplicado!
    "PT-06",                               # PT-06 duplicado!
]

print(f"Total de registros brutos : {len(pontos_campo)}")
print(f"Registros                 : {pontos_campo}")
print()

pontos_unicos = sorted(set(pontos_campo))   # set remove duplicatas
print(f"Pontos unicos             : {len(pontos_unicos)}")
print(f"Lista limpa               : {pontos_unicos}")
````

## 📖 BLOCO 4 — Dicionários
### ⏱️ ~20 minutos

Um **dicionário** armazena pares **chave: valor**.

```python
dicionario = {"chave1": valor1, "chave2": valor2}
```

- **Chaves**: únicas, imutáveis (str, int, float, tupla)
- **Valores**: qualquer tipo, inclusive outras listas ou dicionários
- Acesso por chave: `dic["chave"]`
- Busca extremamente rápida (usa hash table)

> **Analogia cartográfica:** um dicionário é como uma **tabela de atributos** do SIG
> onde cada coluna (atributo) tem um nome (chave) e um valor por feature.


````python
# ── 4.1  Criacao e acesso basico ──────────────────────────────────────────────

# Metadados de uma imagem de satelite (similar a cabecalho de arquivo GeoTIFF)
imagem = {
    "satelite"   : "Sentinel-2A",
    "data"       : "2024-07-15",
    "hora_utc"   : "13:42:18",
    "resolucao"  : 10,              # metros
    "bandas"     : 13,
    "datum"      : "WGS84",
    "fuso_utm"   : 23,
    "cobertura_nuvem": 4.7,         # percentual
    "processado" : True,
}

print("=== Metadados da Imagem ===")
print(f"Satelite      : {imagem['satelite']}")
print(f"Data          : {imagem['data']}")
print(f"Resolucao     : {imagem['resolucao']} m")
print(f"Cobertura     : {imagem['cobertura_nuvem']}%")
print(f"Datum         : {imagem['datum']}")
print()
print(f"Total de atributos: {len(imagem)}")
````

````python
# ── 4.2  Adicionar, atualizar e remover entradas ──────────────────────────────

# Adicionar nova chave
imagem["operador"] = "Ana Costa"
print("Chave 'operador' adicionada:", imagem["operador"])

# Atualizar valor existente
imagem["cobertura_nuvem"] = 3.2
print("Cobertura atualizada       :", imagem["cobertura_nuvem"])

# Remover chave com pop (retorna o valor)
hora = imagem.pop("hora_utc")
print("'hora_utc' removida, valor  :", hora)

# del — remove sem retornar
del imagem["processado"]
print("Chave 'processado' removida.")
print()
print("Dicionario atual:", list(imagem.keys()))
````

````python
# ── 5.5  APLICACAO FINAL: GeoJSON simplificado manualmente ──────────────────
# GeoJSON e o formato padrao da web para dados geoespaciais.
# Sua estrutura e um dicionario aninhado — vamos criar um do zero!

import json, math

def calcular_fuso(lon):
    return int((lon + 180) / 6) + 1

# Estacoes GNSS como features GeoJSON
features = []
for sigla, dados in estacoes_rbmc.items():
    feature = {
        "type"      : "Feature",
        "geometry"  : {
            "type"       : "Point",
            "coordinates": [dados["lon"], dados["lat"]]
        },
        "properties": {
            "sigla"  : sigla,
            "cidade" : dados["cidade"],
            "alt_m"  : dados["alt"],
            "fuso"   : calcular_fuso(dados["lon"]),
            "hemi"   : "S" if dados["lat"] < 0 else "N",
        }
    }
    features.append(feature)

geojson = {
    "type"    : "FeatureCollection",
    "crs"     : {"type": "name", "properties": {"name": "EPSG:4674"}},
    "features": features
}

# Salvar em arquivo
with open("/home/sandbox/estacoes_rbmc.geojson", "w", encoding="utf-8") as f:
    json.dump(geojson, f, ensure_ascii=False, indent=2)

print("GeoJSON gerado com sucesso!")
print(f"Features: {len(geojson['features'])}")
print()
print("Estrutura de uma feature:")
print(json.dumps(features[0], indent=4, ensure_ascii=False))
````

````python
# ── Gabarito — Exercicio 2 ────────────────────────────────────────────────────
import math

vertices = [
    (685100.0, 7460200.0),
    (685340.0, 7460200.0),
    (685340.0, 7459950.0),
    (685100.0, 7459950.0),
]

# a) e b) Perimetro
perimetro = 0.0
n = len(vertices)
for i in range(n):
    e1, n1 = vertices[i]
    e2, n2 = vertices[(i + 1) % n]   # indice circular
    perimetro += math.sqrt((e2-e1)**2 + (n2-n1)**2)

# c) Area pela formula de Gauss (Shoelace)
area = 0.0
for i in range(n):
    xi   = vertices[i][0]
    yi_p = vertices[(i + 1) % n][1]
    yi_m = vertices[(i - 1) % n][1]
    area += xi * (yi_p - yi_m)
area = abs(area) / 2

print(f"Perimetro : {perimetro:.2f} m")
print(f"Area      : {area:.2f} m2  |  {area/10000:.4f} ha")
````

````python
# ── Gabarito — Exercicio Final (versao base) ─────────────────────────────────
import math, matplotlib.pyplot as plt
import matplotlib.patches as mpatch

def shoelace(verts):
    n = len(verts)
    area = 0.0
    for i in range(n):
        xi   = verts[i][0]
        yi_p = verts[(i+1) % n][1]
        yi_m = verts[(i-1) % n][1]
        area += xi * (yi_p - yi_m)
    return abs(area) / 2

cadastro = {
    "responsavel": "Equipe Cartografia UFPR",
    "data"       : "2024-09-10",
    "datum"      : "SIRGAS2000",
    "fuso"       : 22,
    "municipio"  : "Curitiba",
}

lotes = [
    {
        "id"          : "LOTE-001",
        "proprietario": "Jose Silva",
        "vertices"    : [(1000.0,1000.0),(1040.0,1000.0),(1040.0,1025.0),(1000.0,1025.0)],
        "usos"        : {"residencial", "garagem"},
    },
    {
        "id"          : "LOTE-002",
        "proprietario": "Maria Oliveira",
        "vertices"    : [(1040.0,1000.0),(1090.0,1000.0),(1090.0,1030.0),(1040.0,1030.0)],
        "usos"        : {"comercial", "residencial"},
    },
    {
        "id"          : "LOTE-003",
        "proprietario": "Carlos Costa",
        "vertices"    : [(1000.0,1025.0),(1040.0,1025.0),(1090.0,1030.0),(1000.0,1060.0)],
        "usos"        : {"residencial"},
    },
]

# Calcular areas
for lote in lotes:
    lote["area_m2"] = shoelace(lote["vertices"])

# Relatorio
print("=== Relatorio Cadastral ===")
print(f"Municipio: {cadastro['municipio']}  |  Datum: {cadastro['datum']}  |  Fuso: {cadastro['fuso']}")
print()
print(f"{'Lote':<10} {'Proprietario':<20} {'Area (m2)':>10} {'Usos'}")
print("-" * 60)
for lote in lotes:
    usos_str = ", ".join(sorted(lote["usos"]))
    print(f"{lote['id']:<10} {lote['proprietario']:<20} {lote['area_m2']:>10.2f}  {usos_str}")

print()
total_area = sum(l["area_m2"] for l in lotes)
print(f"Area total: {total_area:.2f} m2  ({total_area/10000:.4f} ha)")

# Visualizacao dos lotes
fig, ax = plt.subplots(figsize=(8, 6))
cores_lote = ["#3498db", "#e74c3c", "#2ecc71"]
for lote, cor in zip(lotes, cores_lote):
    verts = lote["vertices"] + [lote["vertices"][0]]
    xs = [v[0] for v in verts]
    ys = [v[1] for v in verts]
    ax.fill(xs[:-1], ys[:-1], color=cor, alpha=0.4)
    ax.plot(xs, ys, color=cor, linewidth=1.5)
    cx = sum(v[0] for v in lote["vertices"]) / len(lote["vertices"])
    cy = sum(v[1] for v in lote["vertices"]) / len(lote["vertices"])
    ax.text(cx, cy, f"{lote['id']} - {lote['area_m2']:.0f} m2",
            ha="center", va="center", fontsize=8, fontweight="bold")

ax.set_xlabel("Este (m)"); ax.set_ylabel("Norte (m)")
ax.set_title("Planta de Lotes — Levantamento Cadastral", fontweight="bold")
ax.set_aspect("equal"); ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig("/home/sandbox/planta_cadastral.png", dpi=130, bbox_inches="tight")
plt.show()
print("Planta salva: planta_cadastral.png")
````
