# Tarefa
Crie um programa em Python que escreva por extenso números de 0 a 9999.

## Definir as palavras para cada número

### subtarefa:
Criar dicionários ou listas para armazenar as palavras correspondentes aos números de 0 a 19, dezenas (20, 30, etc.), centenas (100, 200, etc.) e milhares.


**Reasoning**:
Create dictionaries and a variable to store the words corresponding to numbers, tens, hundreds, and thousands as specified in the instructions.



```python
unidades = {
    0: "zero", 1: "um", 2: "dois", 3: "três", 4: "quatro", 5: "cinco", 6: "seis",
    7: "sete", 8: "oito", 9: "nove", 10: "dez", 11: "onze", 12: "doze", 13: "treze",
    14: "quatorze", 15: "quinze", 16: "dezesseis", 17: "dezessete", 18: "dezoito", 19: "dezenove"
}

dezenas = {
    20: "vinte", 30: "trinta", 40: "quarenta", 50: "cinquenta", 60: "sessenta",
    70: "setenta", 80: "oitenta", 90: "noventa"
}

centenas = {
    100: "cem", 200: "duzentos", 300: "trezentos", 400: "quatrocentos", 500: "quinhentos",
    600: "seiscentos", 700: "setecentos", 800: "oitocentos", 900: "novecentos"
}

mil = "mil"
```

## Criar uma função para escrever números menores que 20

### Subtarefa:
Desenvolver uma função que receba um número entre 0 e 19 e retorne sua representação por extenso.


**Reasoning**:
Define a function that takes a number as input and returns its Portuguese representation by accessing the `unidades` dictionary.



```python
def numero_por_extenso_0_a_19(numero):
  """
  Converte um número entre 0 e 19 para sua representação por extenso em português.

  Args:
    numero: O número a ser convertido (int).

  Returns:
    A representação por extenso do número (str).
  """
  if 0 <= numero <= 19:
    return unidades[numero]
  else:
    return "Número fora do intervalo permitido (0-19)"
```

## Criar uma função para escrever números menores que 100

### Subtask:
Desenvolver uma função que receba um número entre 20 e 99 e retorne sua representação por extenso, utilizando a função anterior para as unidades, se necessário.


**Reasoning**:
Define the function `numero_por_extenso_20_a_99` as instructed, handling the range check, separation into tens and units, and combining the parts with "e" where necessary.



```python
def numero_por_extenso_20_a_99(numero):
  """
  Converte um número entre 20 e 99 para sua representação por extenso em português.

  Args:
    numero: O número a ser convertido (int).

  Returns:
    A representação por extenso do número (str).
  """
  if 20 <= numero <= 99:
    dezena = (numero // 10) * 10
    unidade = numero % 10

    extenso = dezenas[dezena]

    if unidade > 0:
      extenso += " e " + unidades[unidade]

    return extenso
  else:
    return "Número fora do intervalo permitido (20-99)"
```

## Criar uma função para escrever números menores que 1000

### Subtarefa:
Desenvolver uma função que receba um número entre 100 e 999 e retorne sua representação por extenso, utilizando as funções anteriores para as dezenas e unidades, se necessário.


> **Raciocínio**:
> Defina a função `numero_por_extenso_100_a_999` para converter números entre 100 e 999 para sua representação em português, seguindo as instruções fornecidas para lidar com centenas, dezenas e unidades.



```python
def numero_por_extenso_100_a_999(numero):
  """
  Converte um número entre 100 e 999 para sua representação por extenso em português.

  Args:
    numero: O número a ser convertido (int).

  Returns:
    A representação por extenso do número (str).
  """
  if not (100 <= numero <= 999):
    return "Número fora do intervalo permitido (100-999)"

  centena = (numero // 100) * 100
  resto = numero % 100

  extenso = ""

  if centena == 100 and resto == 0:
      extenso = "cem"
  else:
      extenso = centenas[centena]
      if resto > 0:
          extenso += " e "
          if resto < 20:
              extenso += unidades[resto]
          else:
              extenso += numero_por_extenso_20_a_99(resto)

  return extenso
```

## Criar uma função para escrever números menores que 10000

### Subtask:
Desenvolver uma função que receba um número entre 1000 e 9999 e retorne sua representação por extenso, utilizando as funções anteriores para as centenas, dezenas e unidades, se necessário.


> **Raciocínio:**
> Defina a função numero_por_extenso_1000_a_9999 e implemente a lógica para converter números entre 1000 e 9999 para sua forma extensiva, manipulando a parte dos milhares e a parte restante usando funções e variáveis ​​previamente definidas.



```python
def numero_por_extenso_1000_a_9999(numero):
    """
    Converte um número entre 1000 e 9999 para sua representação por extenso em português.

    Args:
      numero: O número a ser convertido (int).

    Returns:
      A representação por extenso do número (str).
    """
    if not (1000 <= numero <= 9999):
        return "Número fora do intervalo permitido (1000-9999)"

    milhares = numero // 1000
    resto = numero % 1000

    extenso = unidades[milhares] + " " + mil

    if resto > 0:
        extenso += " e "
        if resto < 20:
            extenso += numero_por_extenso_0_a_19(resto)
        elif resto < 100:
            extenso += numero_por_extenso_20_a_99(resto)
        else:
            extenso += numero_por_extenso_100_a_999(resto)

    return extenso
```

## Implementar a lógica principal

### Subtask:
Chamar as funções apropriadas com base no número de entrada.


**Reasoning**:
Create the main function `escrever_numero_por_extenso` that checks the input number range and calls the appropriate helper functions based on the range.



```python
def escrever_numero_por_extenso(numero):
    """
    Converte um número entre 0 e 9999 para sua representação por extenso em português.

    Args:
      numero: O número a ser convertido (int).

    Returns:
      A representação por extenso do número (str) ou uma mensagem de erro.
    """
    if 0 <= numero <= 19:
        return numero_por_extenso_0_a_19(numero)
    elif 20 <= numero <= 99:
        return numero_por_extenso_20_a_99(numero)
    elif 100 <= numero <= 999:
        return numero_por_extenso_100_a_999(numero)
    elif 1000 <= numero <= 9999:
        return numero_por_extenso_1000_a_9999(numero)
    else:
        return "Número fora do intervalo permitido (0-9999)"
```
