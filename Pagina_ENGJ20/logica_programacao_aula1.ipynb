
- Indexação começa em **0**: `coordenadas[0]` é o primeiro elemento

````python
# ── Listas de coordenadas UTM ─────────────────────────────────────────────
# Pontos levantados em um levantamento planimétrico (UTM – fuso 23S)

pontos_x = [683_450.0, 683_920.5, 684_100.3, 683_750.8, 683_300.2]
pontos_y = [7_456_200.0, 7_456_650.3, 7_456_300.7, 7_455_900.1, 7_455_700.5]
nomes    = ["PT-01", "PT-02", "PT-03", "PT-04", "PT-05"]

print(f"{'Ponto':<8} {'Este (m)':<16} {'Norte (m)':<16}")
print("-" * 42)
for i in range(len(nomes)):
    print(f"{nomes[i]:<8} {pontos_x[i]:<16.2f} {pontos_y[i]:<16.2f}")
````

````python
# ── enumerate() — forma idiomática do Python ──────────────────────────────

print("Relatório de pontos:")
for idx, nome in enumerate(nomes):
    print(f"  [{idx+1}] {nome}  E={pontos_x[idx]:.2f} m  N={pontos_y[idx]:.2f} m")
````

````python
# ── Distâncias entre pontos consecutivos ──────────────────────────────────
import math

print("Distâncias entre pontos consecutivos:")
distancia_total = 0.0

for i in range(len(nomes) - 1):
    dx = pontos_x[i+1] - pontos_x[i]
    dy = pontos_y[i+1] - pontos_y[i]
    d  = math.sqrt(dx**2 + dy**2)
    distancia_total += d
    print(f"  {nomes[i]} -> {nomes[i+1]} : {d:.3f} m")

print()
print(f"Comprimento total do trajeto: {distancia_total:.3f} m")
````


## 🔧 BLOCO 5 — Funções e Aplicação Final
### ⏱️ ~20 minutos

**Funções** encapsulam um bloco de código reutilizável.
Princípio: **escreva uma vez, use várias vezes.**

```python
def nome_da_funcao(param1, param2):
    # código
    return resultado
```

- `def` → define a função
- `return` → devolve o resultado
- **Parâmetros com valor padrão** evitam repetição: `def f(x, y=0):`


````python
# ── Função: GMS → Graus Decimais ─────────────────────────────────────────

def gms_para_decimal(graus, minutos, segundos, direcao="S"):
    '''Converte coordenadas de GMS para Graus Decimais.

    Args:
        graus    (int)   : graus inteiros
        minutos  (int)   : minutos inteiros
        segundos (float) : segundos decimais
        direcao  (str)   : N, S, E ou W

    Returns:
        float: coordenada em graus decimais (negativo para S e W)
    '''
    decimal = graus + minutos / 60 + segundos / 3600
    if direcao in ("S", "W"):
        decimal = -decimal
    return decimal


# Testando a função
lat_sp  = gms_para_decimal(23, 32, 48.5, "S")
lon_sp  = gms_para_decimal(46, 38,  9.6, "W")
lat_bsb = gms_para_decimal(15, 47, 51.0, "S")
lon_bsb = gms_para_decimal(47, 53, 31.3, "W")

print(f"Sao Paulo  -> Lat: {lat_sp:.6f}  Lon: {lon_sp:.6f}")
print(f"Brasilia   -> Lat: {lat_bsb:.6f}  Lon: {lon_bsb:.6f}")
````

````python
# ── Função: Distância Haversine ───────────────────────────────────────────
# Calcula a distância entre dois pontos na superfície terrestre.
# Amplamente usada em GPS e navegação.

import math

def distancia_haversine(lat1, lon1, lat2, lon2, raio_terra=6_371_000):
    '''Calcula a distância geodésica usando a fórmula de Haversine.

    Args:
        lat1, lon1 (float): coordenadas do ponto 1 em graus decimais
        lat2, lon2 (float): coordenadas do ponto 2 em graus decimais
        raio_terra (float): raio médio da Terra em metros

    Returns:
        float: distância em metros
    '''
    lat1, lon1, lat2, lon2 = map(math.radians, [lat1, lon1, lat2, lon2])
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = math.sin(dlat/2)**2 + math.cos(lat1) * math.cos(lat2) * math.sin(dlon/2)**2
    c = 2 * math.asin(math.sqrt(a))
    return raio_terra * c


# Sao Paulo -> Brasilia
dist_m  = distancia_haversine(lat_sp, lon_sp, lat_bsb, lon_bsb)
dist_km = dist_m / 1000

print(f"Distancia Sao Paulo -> Brasilia")
print(f"  {dist_m:,.0f} m")
print(f"  {dist_km:,.1f} km")
````

````python
# ── Função: Calcular fuso UTM ─────────────────────────────────────────────

def calcular_fuso_utm(longitude, latitude):
    '''Retorna o fuso UTM e o hemisfério a partir das coordenadas.'''
    fuso = int((longitude + 180) / 6) + 1
    hemisferio = "N" if latitude >= 0 else "S"
    return fuso, hemisferio


# Teste
f, h = calcular_fuso_utm(lon_bsb, lat_bsb)
print(f"Brasilia -> Fuso {f}{h}")
````

````python
# ── Gabarito – Exercício Final ────────────────────────────────────────────

def decimal_para_gms(dd):
    '''Converte graus decimais para (graus, minutos, segundos, direcao).'''
    direcao = "N" if dd >= 0 else "S"
    dd_abs  = abs(dd)
    g = int(dd_abs)
    m = int((dd_abs - g) * 60)
    s = ((dd_abs - g) * 60 - m) * 60
    return g, m, s, direcao

def decimal_para_gms_lon(dd):
    '''Igual à anterior mas para longitude (E/W).'''
    direcao = "E" if dd >= 0 else "W"
    dd_abs  = abs(dd)
    g = int(dd_abs)
    m = int((dd_abs - g) * 60)
    s = ((dd_abs - g) * 60 - m) * 60
    return g, m, s, direcao

def relatorio_ponto(nome, lat, lon, alt):
    '''Exibe um relatorio geodesico formatado de um ponto.'''
    lg, lm, ls, ld = decimal_para_gms(lat)
    og, om, os_,od = decimal_para_gms_lon(lon)
    fuso, hemi     = calcular_fuso_utm(lon, lat)

    print("-" * 48)
    print(f" PONTO     : {nome}")
    print(f" Latitude  : {lat:>11.6f}   ({lg}d {lm:02d}' {ls:05.2f}  {ld})")
    print(f" Longitude : {lon:>11.6f}   ({og}d {om:02d}' {os_:05.2f}  {od})")
    print(f" Altitude  : {alt:>9.2f} m")
    print(f" Fuso UTM  :  {fuso}{hemi}")
    print("-" * 48)


# Testes
relatorio_ponto("PT-07", -23.550520, -46.633309, 760.0)
relatorio_ponto("PT-08", lat_bsb, lon_bsb, 1172.0)
````
